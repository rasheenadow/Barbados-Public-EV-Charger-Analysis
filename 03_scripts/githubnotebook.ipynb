## Push to github

```python
# # Example Python setup to push to GitHub
import os

# Set working directory
os.chdir("/home/user/capstone")
print("Working directory:", os.getcwd())

# Git identity (user-defined)
os.system('git config --global user.name "YOUR_USERNAME"')
os.system('git config --global user.email "YOUR_EMAIL"')


# Initialize repo (run once)
os.system("git init")

# Add remote (user-specific)
repo_url = "https://github.com/USERNAME/REPOSITORY.git"
os.system(f"git remote add origin {repo_url}")

# Stage changes
os.system("git add -A")

# Commit
os.system('git commit -m "Reproducible commit: capstone project update"')

# Push (authentication handled separately)
os.system("git branch -M main")
os.system("git push -u origin main")
```
